**#MARKDOWN Description**

1.The Dataset represents data of an online product company with supplies such as PlugTech which sells refurbished Data.


2. -Customer Segmentation: Managers can use Cluster Analysis to group customers based on which Aisles or Departments they frequent, allowing for targeted promotional campaigns
-Departmental Staffing: Descriptive analytics on Orders can highlight peak times for specific Departments, helping managers determine optimal staffing levels to handle order surges
-Demand Forecasting: The company can use time-series analysis on prior orders to forecast consumption of fast-moving products, ensuring the right amount of inventory is stocked to avoid stock-outs

3. Descriptive analytics:
   ORDERS Table, PRODUCTS, AISLES,ORDER_PRODUCTS_PRIOR Table and DEPARTMENTS Tables

   Predictive Analytics:
   ORDER_PRODUCTS_PRIOR and ORDER_PRODUCTS_TRAIN Tables

   Prescriptive Analytics:
   ORDERS, PRODUCTS, AISLES, DEPARTMENTS



1.   CORE TRANSACTION LINK (orders ↔ order_products)

 Orders table connects to ORDER_PRODUCTS_PRIOR and ORDER_PRODUCTS_TRAIN via order_id. This creates a one-to many relationship where each order can contain multiple products. Together, these three tables form the complete   transaction history.

2.   PRODUCT IDENTIFICATION (order_products → products):

 product_id in both ORDER_PRODUCTS_PRIOR and ORDER_PRODUCTS_TRAIN tables links to the PRODUCTS table. This allows retrieval of product details (name, aisle, department) for any item purchased in an order.

3.   PRODUCT CATEGORIZATION (products → aisles & departments)

   Each product has two categorical links:
   * aisle_id connects to AISLES table (134 shopping locations)

   * department_id connects to DEPARTMENTS table (21 departments)

   These create a two-level hierarchy for product classification.
